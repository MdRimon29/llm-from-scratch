### **Create tensors**

##### In PyTorch, a tensor is a multidimensional array (similar to a NumPy array) that stores numerical data and supports GPU acceleration and automatic differentiation.

```python
import torch
tensor0d = torch.tensor(1)  # zero-dimensional tensor(scalar)
tensor1d = torch.tensor([1, 2, 3])  # one-dimensional tensor(vector)
tensor2d = torch.tensor([[1, 2],    # two-dimensional tensor(matrix)
                         [3, 4]]) 
tensor3d = torch.tensor([[[1, 2], [3, 4]],  # four-dimensional tensor
                         [[5, 6], [7, 8]]]) 
```

```python
tensor3d.dtype  # In PyTorch, default 64-bit integer datatype
```

#### **Tensor dtypes**

```python
tensor1d = torch.tensor([1, 2, 3])  # In PyTorch, default 64-bit integer datatype
print(tensor1d.dtype)
```

```python
floatvec = torch.tensor([1.0, 2.0, 3.0])    # default 32-bit float datatype
print(floatvec.dtype)
```

```python
floatvec = tensor1d.to(torch.float32)   # change precision using .to() method
print(floatvec.dtype)
```

#### **Common PyTorch tensor operations**

```python
tensor2d = torch.tensor([[1, 2, 3],     # create new tensors
                         [4, 5, 6]])
print(tensor2d)
```

```python
print(tensor2d.shape)   # shape of a tensor
```

```python
print(tensor2d.reshape(3, 2))   # reshape the tensor
```

```python
print(tensor2d.view(3, 2))  # reshape the tensor(another way)
```

```python
print(tensor2d.T)   # transpose a tensor
```

```python
print(tensor2d.matmul(tensor2d.T))  # multiply two matrices in pytorch
```

```python
print(tensor2d @ tensor2d.T)    # multiply two metrices in another way(by using @ operator)
```
